# Q13: scan_v2_right_side_5 portfolio backtest

Locked portfolio-level comparison of all bull flips, formal breakout, pullback, their union, simple SuperTrend, and SPY buy-and-hold. The production decision stream is treated as point-in-time input; all trades execute on the next symbol session.


```python
import os
from pathlib import Path

def find_repo_root(start=Path.cwd()):
    for candidate in [start, *start.parents]:
        if (candidate / 'pyproject.toml').exists() and (candidate / 'src' / 'xquant').exists():
            return candidate
    raise RuntimeError('repository root not found')

REPO_ROOT = find_repo_root()
os.chdir(REPO_ROOT)
OUTPUT_DIR = REPO_ROOT / 'outputs' / 'scan-v2-right-side-5-portfolio'
```

## 1. Reproduce locked backtest

Base assumptions are pre-registered in the script: CNY 1,000,000 initial capital, 5 bps commission and 5 bps adverse slippage per side, 10% per position, 10 concurrent positions, and 100% gross exposure. Sensitivity tables are diagnostics, not parameter selection.


```python
import subprocess

subprocess.run([
    'uv', 'run', 'python', 'scripts/research_scan_v2_right_side_5_portfolio.py',
    '--decisions', 'outputs/unified-scan-policy-v2-breakout-r5/decision_stream.parquet',
    '--decision-metadata', 'outputs/unified-scan-policy-v2-breakout-r5/decision_stream.metadata.json',
    '--price-data-dir', '../trading/backend/data',
    '--output-dir', str(OUTPUT_DIR),
], check=True)
```
